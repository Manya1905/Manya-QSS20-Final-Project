# 03 - Analysis

Loads the cleaned trend DataFrames and USDA subsets saved by `02_data_cleaning.ipynb`, generates all charts (peak-month comparison, per-trend monthly frequency, USDA retail sales), and exports all generated PNGs into a single zip file.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import zipfile
```

## Loading data from previous notebook

```python
feta_pasta_df = pd.read_csv('../cleaned_data/feta_pasta.csv')
sourdough_df = pd.read_csv('../cleaned_data/sourdough.csv')
banana_bread_df = pd.read_csv('../cleaned_data/banana_bread.csv')
baked_oats_df = pd.read_csv('../cleaned_data/baked_oats.csv')
dalgona_coffee_df = pd.read_csv('../cleaned_data/dalgona_coffee.csv')

flour = pd.read_csv('../cleaned_data/usda_flour_and_mixes.csv')
sweet_mixes = pd.read_csv('../cleaned_data/usda_sweet_mixes.csv')

# 'month' and 'Date' need to be reconstructed as proper date types after the CSV round-trip
for d in [feta_pasta_df, sourdough_df, banana_bread_df, baked_oats_df, dalgona_coffee_df]:
    d["date_parsed"] = pd.to_datetime(d["date_parsed"], utc=True)
    d["month"] = d["date_parsed"].dt.to_period("M")

flour["Date"] = pd.to_datetime(flour["Date"])
sweet_mixes["Date"] = pd.to_datetime(sweet_mixes["Date"])

sourdough_df.shape
```

## Instagram trend graphs

```python
peak_data = {}
for name, df in [("Feta Pasta", feta_pasta_df), ("Sourdough", sourdough_df),
                  ("Banana Bread", banana_bread_df), ("Baked Oats", baked_oats_df),
                  ("Dalgona Coffee", dalgona_coffee_df)]:
    peak_month = df.groupby("month").size().idxmax()
    peak_data[name] = peak_month

# Convert to something plottable — months as a numeric/ordered value
peak_series = pd.Series(peak_data).apply(lambda p: p.to_timestamp())

plt.figure(figsize=(10, 5))
plt.scatter(peak_series.values, peak_series.index, s=150)
plt.title("Peak Month by Trend")
plt.xlabel("Month")
plt.ylabel("Trend")
plt.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.savefig("instagram_baseline_total_volume.png", dpi=150)
plt.show()
```

```python
all_trends = {
    "feta_pasta": feta_pasta_df,
    "sourdough": sourdough_df,
    "banana_bread": banana_bread_df,
    "baked_oats": baked_oats_df,
    "dalgona_coffee": dalgona_coffee_df,
}

for name, d in all_trends.items():
    monthly_split = d.groupby(["month", "is_covid_framed"]).size().unstack(fill_value=0)
    monthly_split.plot(kind="bar", stacked=True, figsize=(12, 5))
    plt.title(f"Posts per Month: {name.replace('_', ' ').title()}")
    plt.xlabel("Month"); plt.ylabel("Number of Posts")
    plt.legend(["Not COVID-framed", "COVID-framed"])
    plt.xticks(rotation=45); plt.tight_layout()
    plt.savefig(f"instagram_{name}_monthly_frequency.png", dpi=150)
    plt.show()
```

## Generating Graphics for Food Sales Data

```python
plt.figure(figsize=(12, 5))
plt.plot(flour["Date"], flour["Value"] / 1e6, marker="o", markersize=3, color="#6b4226")
plt.axvline(pd.Timestamp("2020-03-13"), color="red", linestyle="--", alpha=0.7,
            label="US National Emergency Declared (3/13/20)")
plt.title("Weekly U.S. Retail Sales: Flour and Mixes (Dec 2019 - Jan 2021)")
plt.xlabel("Week")
plt.ylabel("Sales ($ Millions)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("viz1_flour_sales.png", dpi=150)
plt.show()
```

```python
plt.figure(figsize=(12, 5))
plt.plot(sweet_mixes["Date"], sweet_mixes["Value"] / 1e6, marker="o", markersize=3, color="#c9a66b")
plt.axvline(pd.Timestamp("2020-03-13"), color="red", linestyle="--", alpha=0.7,
            label="US National Emergency Declared (3/13/20)")
plt.title("Weekly U.S. Retail Sales: Sweet Mixes (Pancake, Muffin, Cake) (Dec 2019 - Jan 2021)")
plt.xlabel("Week")
plt.ylabel("Sales ($ Millions)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("viz2_sweet_mixes_sales.png", dpi=150)
plt.show()
```

## Exporting Graphs

```python
png_files = [f for f in os.listdir() if f.endswith(".png")]
print("Found", len(png_files), "PNG files:", png_files)

with zipfile.ZipFile("all_project_visualizations.zip", "w") as zf:
    for f in png_files:
        zf.write(f)

print("Zipped into all_project_visualizations.zip")
```
